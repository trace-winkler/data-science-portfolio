```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

df2 = pd.read_csv(r"C:\Users\ncwin_05165p4\OneDrive\Desktop\Data Science 2\Project 1 Stuff\Project 1 Info2 DTSC CSV.csv")

for col in ['one_high_percentage', 'two_high_percentage']:
    df2[col] = df2[col].str.rstrip('%').astype(float)

plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(12, 5))
c_one = "#5FB0CB"
c_two = '#FFA500'
ax.plot(df2['year'], df2['one_high_percentage'], marker='o', markersize=6,
        linewidth=2.5, label='One-High Safety Usage', color=c_one, zorder=3)
ax.plot(df2['year'], df2['two_high_percentage'], marker='o', markersize=6,
        linewidth=2.5, label='Two-High Safety Usage', color=c_two, zorder=3)

for series, color, offset in [
    ('one_high_percentage', c_one, 10),
    ('two_high_percentage', c_two, -16),
]:
    for idx in [0, len(df2) - 1]:
        x = df2['year'].iloc[idx]
        y = df2[series].iloc[idx]
        ax.annotate(f'{y:.1f}%', (x, y), textcoords="offset points",
                    xytext=(0, offset), ha='center', fontsize=10,
                    fontweight='bold', color=color)

ax.set_title('NFL Safety Usage, 2015-2026', fontsize=18, fontweight='bold', pad=18)
ax.text(0.5, 1.03, 'Trend in One-High vs. Two-High Coverage Shells',
        transform=ax.transAxes, ha='center', fontsize=11, color='#555555')
ax.set_xlim(2014, 2027)
ax.set_ylim(0, 80)
ax.yaxis.set_major_formatter(PercentFormatter(xmax=100))
ax.set_xticks(range(2015, 2027, 1))
ax.set_xlabel('Year', fontsize=12, fontweight='bold', labelpad=10)
ax.set_ylabel('Defensive Percentage', fontsize=12, fontweight='bold', labelpad=10)
ax.legend(loc='lower center', bbox_to_anchor=(0.5, -0.22), ncol=2,
          frameon=False, fontsize=11)
ax.grid(True, axis='y', alpha=0.3, linestyle='--')
ax.grid(False, axis='x')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#888888')
ax.spines['bottom'].set_color('#888888')
```
